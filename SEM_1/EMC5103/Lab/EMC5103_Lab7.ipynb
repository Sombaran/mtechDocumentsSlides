```python
# Generate a random integer from interval (a,b)
import numpy as np
rng=np.random.default_rng()

a=-10
b=10
random_integers=rng.integers(a,b,size=10)
print(f"The random integer from {a,b} {random_integers}")
```

```python
# Generate a random float from interval (a,b)
a=5
b=15
random_float=rng.random(size=10)*(b-a)+a

print(f"The random float from {a,b} {random_float}")
```

```python
# Generate random sample from exponential distribution
import numpy as np

rng=np.random.default_rng()

def exponential_sample(lam,n):
    R=rng.random(n)

    return (-1/lam)*np.log(1-R)

lam=0.1
n=10
random_sample_exp=exponential_sample(lam,n)
print(f"Random sample from exponential distribution is {random_sample_exp}")
```

```python
# Generate random sample from uniform distribution
import numpy as np

rng=np.random.default_rng()

def uniform_sample(a,b,n):
    R=rng.random(n)

    return a+R*(b-a)

a=2
b=10
n=10
random_sample_uniform=uniform_sample(a,b,n)
print(f"Random sample from uniform distribution is {random_sample_uniform}")
```

```python
# Using inbuilt function (Exponential)
lam=5
size=10

random_exp=rng.exponential(1/lam,size)
print(random_exp)
```

```python
# Using inbuilt function (uniform)
a=2
b=10
size=10

random_uniform=rng.uniform(a,b,size)
print(random_uniform)
```

```python
# Using inbuilt function (normal distribution)

mu=0
sigma=1
size=10

random_normal=rng.normal(mu,sigma,size)
print(random_normal)
```
