```python
import numpy as np
import matplotlib.pyplot as plt


# Function to plot PDF (probability density function) and CDF (cumulative distribution function)
def plot_conti_dist(dist_name, x_values, pdf_values,cdf_values):
   plt.figure(figsize=(10,4))

   # Plot PDF
   plt.subplot(1,2,1)
   plt.plot(x_values,pdf_values)
   plt.title(f"PDF plot of {dist_name} distribution")
   plt.xlabel("k")
   plt.ylabel("PDF(X=k)")
   plt.grid(True)

   # Plot CDF
   plt.subplot(1,2,2)
   # plt.subplot(number of row, number col, which number of plot yor are ploting out of number of row*number of col)
   plt.plot(x_values,cdf_values)
   plt.title(f"CDF plot of {dist_name} distribution")
   plt.xlabel("k")
   plt.ylabel("P(X<=k)")
   plt.grid(True)
```

```python
from scipy.special import erf

def normal_manual(mu,sigma):
    x=np.linspace(mu-4*sigma,mu+4*sigma,100)

    pdf=(1/(np.sqrt(2*np.pi)*sigma))*np.exp((-(x-mu)**2)/(2*(sigma**2)))

    z=(x-mu)/(np.sqrt(2)*sigma)

    cdf= 0.5*(1+erf(z))

    return x, pdf, cdf
```

```python
mu=70
sigma=10
x,pdf,cdf=normal_manual(mu,sigma)
plot_conti_dist("Normal_Distribution", x, pdf,cdf)
```
