```python
# Sorting of list
a=[2,14,5,6,3,6,9]
a.sort() #Ascending order
print(f"The sorted list is {a}")

# Descending order
b=[12,14,5,9,8,6,10]
b.sort(reverse=True)
print(f"The sorted list is {b}")
```
